```python
# 1. Averaging filter (un-weighted)
image = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Could not read the image.")
    exit()

# Input filter size
filter_size = int(input("Enter filter size (odd number, e.g. 3, 5, 7): "))

if filter_size <= 0 or filter_size % 2 == 0:
    print("Filter size must be a positive odd number.")
    exit()

# Select padding method
print("\nSelect Padding Method:")
print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

choice = int(input("Enter your choice: "))

if choice == 1:
    padding = cv2.BORDER_CONSTANT
elif choice == 2:
    padding = cv2.BORDER_REPLICATE
elif choice == 3:
    padding = cv2.BORDER_REFLECT
else:
    print("Invalid choice.")
    exit()

# Calculate padding size
pad = filter_size // 2

# Apply padding
padded_image = cv2.copyMakeBorder(
    image,
    pad, pad, pad, pad,
    padding,
    value=0
)

# Create averaging kernel
kernel = np.ones(
    (filter_size, filter_size),
    dtype=np.float32
) / (filter_size * filter_size)

# Create output image
output = np.zeros_like(image, dtype=np.float32)

# Apply averaging filter manually
rows, cols = image.shape

for i in range(rows):
    for j in range(cols):

        # Extract neighborhood
        window = padded_image[
            i:i + filter_size,
            j:j + filter_size
        ]

        # Calculate average
        output[i, j] = np.sum(window * kernel)

# Convert to 8-bit image
output = np.clip(output, 0, 255).astype(np.uint8)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Averaging Filter", output)

# Save output
cv2.imwrite("averaging_filter_output.jpg", output)

print("\nAveraging filter applied successfully.")
print("Output saved as: averaging_filter_output.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 2. Weighted filter given by h(x,y)= max(|x|,|y|)
image = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Could not read the image.")
    exit()

# Input filter size
filter_size = int(input("Enter filter size (odd number, e.g. 3, 5, 7): "))

if filter_size <= 0 or filter_size % 2 == 0:
    print("Filter size must be a positive odd number.")
    exit()

# Select padding method
print("\nSelect Padding Method:")
print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

choice = int(input("Enter your choice: "))

if choice == 1:
    padding = cv2.BORDER_CONSTANT
elif choice == 2:
    padding = cv2.BORDER_REPLICATE
elif choice == 3:
    padding = cv2.BORDER_REFLECT
else:
    print("Invalid choice.")
    exit()

# Calculate padding size
pad = filter_size // 2

# Apply padding
padded_image = cv2.copyMakeBorder(
    image,
    pad, pad, pad, pad,
    padding,
    value=0
)

# Create weighted filter kernel
kernel = np.zeros(
    (filter_size, filter_size),
    dtype=np.float32
)

# Generate h(x,y) = max(|x|, |y|)
for x in range(-pad, pad + 1):
    for y in range(-pad, pad + 1):
        kernel[x + pad, y + pad] = max(abs(x), abs(y))

# Normalize kernel
kernel = kernel / np.sum(kernel)

print("\nWeighted Filter Kernel:")
print(kernel)

# Create output image
output = np.zeros_like(image, dtype=np.float32)

# Apply weighted filter manually
rows, cols = image.shape

for i in range(rows):
    for j in range(cols):

        # Extract neighborhood
        window = padded_image[
            i:i + filter_size,
            j:j + filter_size
        ]

        # Apply weighted filter
        output[i, j] = np.sum(window * kernel)

# Convert to 8-bit image
output = np.clip(output, 0, 255).astype(np.uint8)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Weighted Filter", output)

# Save output
cv2.imwrite("weighted_filter_output.jpg", output)

print("\nWeighted filter applied successfully.")
print("Output saved as: weighted_filter_output.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 3. Gaussian filter
image = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Could not read the image.")
    exit()

# Input filter size
filter_size = int(
    input("Enter filter size (odd number, e.g. 3, 5, 7): ")
)

if filter_size <= 0 or filter_size % 2 == 0:
    print("Filter size must be a positive odd number.")
    exit()

# Input sigma
sigma = float(input("Enter sigma value (e.g. 1.0): "))

if sigma <= 0:
    print("Sigma must be greater than 0.")
    exit()

# Select padding method
print("\nSelect Padding Method:")
print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

choice = int(input("Enter your choice: "))

if choice == 1:
    padding = cv2.BORDER_CONSTANT
elif choice == 2:
    padding = cv2.BORDER_REPLICATE
elif choice == 3:
    padding = cv2.BORDER_REFLECT
else:
    print("Invalid choice.")
    exit()

# Calculate padding size
pad = filter_size // 2

# Apply padding
padded_image = cv2.copyMakeBorder(
    image,
    pad, pad, pad, pad,
    padding,
    value=0
)

# ---------------------------------------------------------
# Create Gaussian Kernel
# G(x,y) = exp(-(x^2 + y^2) / (2*sigma^2))
# ---------------------------------------------------------

kernel = np.zeros(
    (filter_size, filter_size),
    dtype=np.float32
)

for x in range(-pad, pad + 1):
    for y in range(-pad, pad + 1):

        kernel[x + pad, y + pad] = np.exp(
            -(x**2 + y**2) / (2 * sigma**2)
        )

# Normalize kernel
kernel = kernel / np.sum(kernel)

print("\nGaussian Filter Kernel:")
print(kernel)

# ---------------------------------------------------------
# Apply Gaussian Filter
# ---------------------------------------------------------

output = np.zeros_like(image, dtype=np.float32)

rows, cols = image.shape

for i in range(rows):
    for j in range(cols):

        # Extract neighborhood
        window = padded_image[
            i:i + filter_size,
            j:j + filter_size
        ]

        # Apply Gaussian kernel
        output[i, j] = np.sum(window * kernel)

# Convert to 8-bit image
output = np.clip(output, 0, 255).astype(np.uint8)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Gaussian Filter", output)

# Save output
cv2.imwrite("gaussian_filter_output.jpg", output)

print("\nGaussian filter applied successfully.")
print("Output saved as: gaussian_filter_output.jpg")

cv2.waitKey(0)
cv2.destroyAllWindows()
```
