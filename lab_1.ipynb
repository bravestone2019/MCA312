```python
# b.  Convert the image into different formats (PNG, TIFF, and many more).

img = cv2.imread(r"C:\Users\Lenovo\OneDrive\Desktop\family\Animesh\01_Thapar_Student_Residencies.jpg")

if img is not None:

    # png 
    cv2.imwrite("Output.png", img)

    # tiff
    cv2.imwrite("Output.tiff", img)

    # bmp
    cv2.imwrite("output.bmp", img)

    print("Image converted to PNG, TIFF and BMP.")

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# c. Perform image resizing.

img = cv2.imread(r"C:\Users\Lenovo\OneDrive\Desktop\family\Animesh\01_Thapar_Student_Residencies.jpg")

if img is not None:
    resized_img = cv2.resize(img, (500, 500))

    cv2.imshow("Resized Image", resized_img)

    cv2.imwrite("resized.jpg", resized_img)

    print("Image resized successfully.")

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# d. Convert a colored image into a grayscale image.

if img is not None:

    gray_img = cv2.cvtcolor(img, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Grayscale Image" gray_img)

    cv2.imwrite("Grayscale.jpg", gray_img)

    print("Image Converted to grayscale.")

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# e.  Read and display a video.

# video = cv2.VideoCapture("video_input.mp4") # For specific file 
video = cv2.VideoCapture(0)

if not video.isOpened():
    print("Error: Could not open video.")
else:
    print("Video opened successfully.")

    while True:

        # Read one frame
        ret, frame = video.read()

        # If video ends
        if not ret:
            break

        # Display frame
        cv2.imshow("Video", frame)

        # Press 'q' to stop
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break

# Release video
video.release()

cv2.destroyAllWindows()
```

```python
# f. Extract frames from a video.
video = cv2.VideoCapture("input.mp4") # Example :- input.mp4 

# Create folder for frames
os.makedirs("frames", exist_ok=True)

frame_number = 0

while True:

    ret, frame = video.read()

    if not ret:
        break

    # Save every frame
    filename = f"frames/frame_{frame_number:04d}.jpg"

    cv2.imwrite(filename, frame)

    frame_number += 1

video.release()

print("Total frames extracted:", frame_number)
```
