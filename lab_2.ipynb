```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image =  r"C:\Users\Lenovo\OneDrive\Desktop\family\Animesh\01_Thapar_Student_Residencies.jpg"
```

```python
# 1. Write a program (WAP) to convert a grayscale image into a binary image.

# Case 1: Assume the mean intensity of the image as the threshold value.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the grayscale image
img = cv2.imread(image, cv2.IMREAD_GRAYSCALE)

# Check if image was loaded
if img is None:
    print("Error: Image not found!")
else:
    # Calculate mean intensity
    threshold = np.mean(img)

    print("Mean Intensity =", threshold)

    # Convert grayscale image to binary image
    binary = np.where(img >= threshold, 255, 0).astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(binary, cmap="gray")
    plt.title(f"Binary Image (Threshold = {threshold:.2f})")
    plt.axis("off")

    plt.show()
```

```python
# Case 2: Input the threshold value from the user.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the grayscale image
img = cv2.imread(image, cv2.IMREAD_GRAYSCALE)

# Check if image was loaded
if img is None:
    print("Error: Image not found!")
else:
    # Take threshold from user
    threshold = int(input("Enter threshold value (0-255): "))

    # Validate threshold
    if threshold < 0 or threshold > 255:
        print("Invalid threshold! Enter a value between 0 and 255.")
    else:
        # Convert grayscale image to binary image
        binary = np.where(img >= threshold, 255, 0).astype(np.uint8)

        # Display images
        plt.figure(figsize=(10, 4))

        plt.subplot(1, 2, 1)
        plt.imshow(img, cmap="gray")
        plt.title("Grayscale Image")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(binary, cmap="gray")
        plt.title(f"Binary Image (Threshold = {threshold})")
        plt.axis("off")

        plt.show()
```

```python

# 2. Write a program (WAP) to convert an RGB (color) image into a grayscale image.
# Case 1: Convert the RGB image to grayscale by taking the mean average of the three color planes (R, G, and B).

# 2. Convert RGB image into grayscale image
# Case 1: Mean average of R, G and B planes

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(image)

# Check if image was loaded
if img is None:
    print("Error: Image not found!")
else:
    # OpenCV reads image as BGR, so convert it to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Separate R, G and B color planes
    R = rgb[:, :, 0]
    G = rgb[:, :, 1]
    B = rgb[:, :, 2]

    # Calculate mean average
    gray = (R + G + B) / 3

    # Convert to uint8
    gray = gray.astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(rgb)
    plt.title("RGB Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image - Mean Average")
    plt.axis("off")

    plt.show()
```

```python
# Case 2: Input the weightages of the three color planes (R, G, and B) from the user. The weightage of each plane should be a value between 0 and 1, and the sum of all three weightages must be equal to 1.

# Case 2: Convert RGB image into grayscale using
# user-defined weightages

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(image)

# Check if image was loaded
if img is None:
    print("Error: Image not found!")
else:
    # OpenCV reads image as BGR, so convert it to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Separate R, G and B planes
    R = rgb[:, :, 0]
    G = rgb[:, :, 1]
    B = rgb[:, :, 2]

    # Input weightages
    while True:
        WR = float(input("Enter weightage for Red (0-1): "))
        WG = float(input("Enter weightage for Green (0-1): "))
        WB = float(input("Enter weightage for Blue (0-1): "))

        # Check individual weightages
        if not (0 <= WR <= 1 and 0 <= WG <= 1 and 0 <= WB <= 1):
            print("Error: Each weightage must be between 0 and 1.")
            continue

        # Check sum
        if not np.isclose(WR + WG + WB, 1):
            print("Error: Sum of all weightages must be equal to 1.")
            print("Current sum =", WR + WG + WB)
            continue

        break

    # Calculate weighted grayscale image
    gray = (WR * R + WG * G + WB * B)

    # Convert to uint8
    gray = gray.astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(rgb)
    plt.title("RGB Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image - Weighted")
    plt.axis("off")

    plt.show()

    # Display weightages
    print("\nWeightages used:")
    print("Red   =", WR)
    print("Green =", WG)
    print("Blue  =", WB)
    print("Sum   =", WR + WG + WB)
```
