```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Import Succesfully")
```

```python
src = r"C:\Users\Lenovo\OneDrive\Desktop\family\Animesh\01_Thapar_Student_Residencies.jpg"

print("Upload Successfully")
```

```python
# Function to add border #
def add_border(image, border_width, border_color):
    """
    image         : Input grayscale/binary image
    border_width  : Width of border in pixels
    border_color  : Color/intensity of border
    """

    if len(image.shape) == 2:
        # Grayscale / Binary image
        bordered_image = cv2.copyMakeBorder(
            image,
            border_width,
            border_width,
            border_width,
            border_width,
            cv2.BORDER_CONSTANT,
            value=border_color
        )

    else:
        # RGB image
        bordered_image = cv2.copyMakeBorder(
            image,
            border_width,
            border_width,
            border_width,
            border_width,
            cv2.BORDER_CONSTANT,
            value=border_color
        )

    return bordered_image

# Read image #

image = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found!")
else:
    # Input border width #
    width = int(input("Enter border width (pixels): "))
    
    # Input border color
    # For grayscale: 0 = Black, 255 = White
    color = int(input("Enter border color (0-255): "))

    if color < 0 or color > 255:
        print("Color must be between 0 and 255.")

    else:
        bordered = add_border(image, width, color)

        # Display #
        plt.figure(figsize=(10, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(image, cmap='gray')
        plt.title("Original Image")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(bordered, cmap='gray')
        plt.title("Image with Border")
        plt.axis("off")

        plt.show()

        # Save result #
        cv2.imwrite("bordered_image.png", bordered)
```

```python
# 2. WAP for Image Compliment. Try this for both binary and grayscale image.

image = cv2.imread(src, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Image not found!")
    exit()

print("\nSelect Image Type:")
print("1. Grayscale")
print("2. Binary")

choice = int(input("Enter choice: "))

# GRAYSCALE IMAGE #

if choice == 1:

    complement = 255 - image

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap="gray")
    plt.title("Original Grayscale")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(complement, cmap="gray")
    plt.title("Grayscale Complement")
    plt.axis("off")

    plt.show()

# BINARY IMAGE #

elif choice == 2:

    # Convert input image to binary #
    _, binary = cv2.threshold(
        image,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Complement #
    complement = 255 - binary

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(binary, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(complement, cmap="gray")
    plt.title("Binary Complement")
    plt.axis("off")

    plt.show()

else:
    print("Invalid choice!")
```
